```python
import os

# Define base project folders
base_dir = "/databricks/driver"
tests_dir = os.path.join(base_dir, "tests")
reports_dir = os.path.join(base_dir, "reports")

# Create directories
os.makedirs(tests_dir, exist_ok=True)
os.makedirs(reports_dir, exist_ok=True)

print("✅ Project structure created:")
print(f"- {base_dir}")
print(f"  ├── tests/")
print(f"  └── reports/")
```

```python
pytest_ini_content = """[pytest]
testpaths = tests
addopts =
    -ra
    --junitxml=reports/junit.xml
    --html=reports/report.html
    --self-contained-html
python_files = test_*.py
"""

with open(os.path.join(base_dir, "pytest.ini"), "w") as f:
    f.write(pytest_ini_content)

print("✅ pytest.ini created at:", os.path.join(base_dir, "pytest.ini"))
```

```python
test_math_code = """def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

def test_addition():
    assert add(2, 3) == 5

def test_multiplication():
    assert multiply(4, 5) == 20

def test_combined_operation():
    result = multiply(add(2, 3), 2)  # (2+3)*2 = 10
    assert result == 10
"""

test_file_path = os.path.join(tests_dir, "test_math.py")
with open(test_file_path, "w") as f:
    f.write(test_math_code)

print("✅ Test file created:", test_file_path)
```
